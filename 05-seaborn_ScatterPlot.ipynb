```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
allDatasets = sns.get_dataset_names()
allDatasets
```

```python
dataset = sns.load_dataset('diamonds')
dataset
#now let's work with less rows of data
smallset = dataset.iloc[0:70]
print(smallset)
```

```python
#NOTE : creating a scatterplot using seaborn between bill_length_mm and bill_depth_mm
sns.scatterplot(
    smallset,     
    y='depth',
    x='price',
    hue='cut'
)
plt.show()
```

## using parameter :    size = column name on the basis of which our data is differenciating
#                   This is use to increase the size of the dots in the scatter plot

#       we can also pass size in a tuple(sizeof(x) , sizeof(Y)) it's like we can specify the size 

```python
sns.scatterplot(
    smallset,     
    y='depth',
    x='price',
    hue='cut',
    size='carat'    # size increasing as carat is increasing 
)
plt.show()
```

# USing sizes=(,) 

```python
sns.scatterplot(
    smallset,     
    y='depth',
    x='price',
    hue='cut',
    sizes=(20, 60)  # 20 for price and 60 for depth
)
plt.show()
```

## NOW using parameter :        s= fixed size of point that is scattring 
#           This doesn't increases or decreses the size on the basis of any parameter 
#           This is always fixed

```python
sns.scatterplot(
    smallset,     
    y='depth',
    x='price',
    hue='cut',
    #size='carat',          This cannot be used together with s=
    s=90        #NOTE size of all dots will be same
)
plt.show()
```

# Using edgecolor=      means giving color to a border
# Using linewidth=      increasing the border-width
# USing style=          for representing different dots with different symbols

```python
sns.scatterplot(
    smallset,     
    y='depth',
    x='price',
    hue='cut',
    #size='carat',          This cannot be used together with s=
    s=90,        #NOTE size of all dots will be same
    edgecolor='black',
    linewidth=2,
    style='cut'
)
plt.show()
```

# USing palette - to change color theme

```python
sns.scatterplot(
    smallset,     
    y='depth',
    x='price',
    hue='cut',
    #size='carat',          This cannot be used together with s=
    s=90,        #NOTE size of all dots will be same
    edgecolor='black',
    linewidth=2,
    style='cut',
    palette='bright'
)
plt.show()
```
