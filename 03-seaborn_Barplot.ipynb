Changing the orientation of the barplots :
    There are two types : 
    a) orient='v'      ---> for vertical
    b) orient='h'      ---> for horizontal

#NOTE  : ❌ Problem

You are using:

x = 'bill_depth_mm' → numeric

y = 'bill_length_mm' → numeric

order = predefined_order → expects a categorical column,
but your y is numeric, not a category like "Dream" or "Torgersen".

Because your plot is horizontal (orient='h'), Seaborn expects:

y → categorical

x → numeric

But currently both are numeric, so order= cannot match anything.

```python

predefined_order = ['Dream', 'Torgersen', 'Biscoe']
predefined_hue_order = ['Female', 'Male']
sns.barplot(
    x='bill_depth_mm',
    y='island',
    data=ds,
    order = predefined_order,
    hue='sex',
    hue_order=predefined_hue_order,
    orient='h'
    )
plt.show()
```

## Changing color of all bars.

NOTE: 🎯 Why does Seaborn ignore color when hue is present?

Because:

color= → means one single color

hue= → needs one color per category, e.g.,

Female = blue

Male = orange

Both cannot work together. So hue wins.

```python
predefined_order = ['Dream', 'Torgersen', 'Biscoe']
predefined_hue_order = ['Female', 'Male']
sns.barplot(
    x='bill_depth_mm',
    y='island',
    data=ds,
    order = predefined_order,
    #hue='sex',      ---> Not using this since color= gives same color for all bars but for(hue) we need multiple
    hue_order=predefined_hue_order,
    orient='h',
    color='g',
    )
plt.show()
```

```python
#NOTE : CASE using (hue=) and (color=) results giving default two different color to each comparing bar and 
# different color to the next bar to which we are comparing 
```

```python
#NOTE : ✅ What is saturation in Seaborn barplot?
'''
saturation controls how intense (vibrant) the bar colors appear.

It adjusts the opacity + color richness of the bars.

✔️ Value range:

saturation = 1 → full color, very vibrant

saturation = 0 → completely faded, almost grey

0 < saturation < 1 → partially faded
'''
```

```python
predefined_order = ['Dream', 'Torgersen', 'Biscoe']
predefined_hue_order = ['Female', 'Male']
sns.barplot(
    x='bill_depth_mm',
    y='island',
    data=ds,
    order = predefined_order,
    hue='sex',      
    hue_order=predefined_hue_order,
    orient='h',
    palette='colorblind',
    saturation=0.7      #---> range of the saturation value is (0 to 1)
    )
plt.show()
```

```python
'''
🔥 ADVANCED QUESTIONS
🚀 Advanced Q1 (ML-style visualization: feature comparison + caps + error bars)

Dataset:

df3 = pd.DataFrame({
    'Model': ['LR', 'LR', 'SVM', 'SVM', 'RF', 'RF'],
    'Dataset': ['Train', 'Test', 'Train', 'Test', 'Train', 'Test'],
    'Accuracy': [88, 84, 92, 89, 96, 93]
})
df3

Question:

Create a barplot comparing Model vs Accuracy with:

y = Model
x = Accuracy
hue='Dataset'
palette='deep'
capsize=0.4 and custom error-bar styling
Show mean accuracy bars with 95% CI
Style: "darkgrid" with context='poster'

Goal: mimic ML model comparison plots (very commonly used in ML papers).
'''

import numpy as np

df3 = pd.DataFrame({
    'Model': ['LR', 'LR', 'SVM', 'SVM', 'RF', 'RF'],
    'Dataset': ['Train', 'Test', 'Train', 'Test', 'Train', 'Test'],
    'Accuracy': [88, 84, 92, 89, 96, 93]
})
df3
sns.set(style='darkgrid', context='poster')
sns.barplot(
        x='Accuracy',
        y='Model',
        data=df3,
        hue='Dataset',
        palette='deep',
        capsize=0.4,
        err_kws={"color":'skyblue', "linewidth":10},
        # The `ci` parameter is deprecated. Use `errorbar=('ci', 95)` for the same effect.
        errorbar=('ci', 95),  # This set the vertical (err bar) for 95% on each bars
        # ci=95,     # This way of doing is deprecated
        estimator=np.mean       # this finds mean for the bar heigth (err_bar)
)
plt.show()
```

```python
'''
🧠 Advanced Q2 (Multiple customizations + ML reasoning)

Dataset:

df4 = pd.DataFrame({
    'Class': ['Good', 'Average', 'Poor', 'Good', 'Average', 'Poor'],
    'Gender': ['M', 'M', 'M', 'F', 'F', 'F'],
    'Score': [90, 75, 50, 92, 72, 48]
})
df4

Question:

Create a barplot that:
Uses horizontal orientation
Orders classes as ['Good', 'Average', 'Poor']
Hue = Gender with hue order ['F', 'M']
Uses a palette from your ML-preferred list ('coolwarm', 'mako', 'Spectral', etc.)
Sets saturation=0.7
Uses err_kws to customize error bar color + line width
Use sns.set(style='ticks', context='notebook')
Add title: "Performance Distribution by Class and Gender"

Bonus reasoning:

Write 2 lines explaining:

Why horizontal barplots help in classification distribution tasks

Why palette selection matters in ML research visualizations
'''
df4 = pd.DataFrame({
    'Class': ['Good', 'Average', 'Poor', 'Good', 'Average', 'Poor'],
    'Gender': ['M', 'M', 'M', 'F', 'F', 'F'],
    'Score': [90, 75, 50, 92, 72, 48]
})
df4

sns.set(style='ticks', context='notebook')
sns.barplot(
    data=df4,
    x='Score',
    y='Class',
    order=['Good', 'Average', 'Poor'],
    hue='Gender',
    hue_order=['F', 'M'],
    palette='Spectral',
    saturation=0.7,
    err_kws={"color":'yellow', "linewidth":9}
)
plt.title('Performance Distribution by Class and Gender')
plt.show()
```
