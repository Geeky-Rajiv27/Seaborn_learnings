```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df = sns.load_dataset('penguins')
df
smallset = df.iloc[0:60]
smallset
```

# 1)   kind='reg' is not supported with hue since regression plot only allows single line but 
#   (hue= ) means multiple groups 

#   2) kind='hex' is also not supported with ( hue = )



```python
sns.set_style('whitegrid')
sns.set_context("talk")
sns.jointplot(
    data=smallset,
    x='bill_length_mm',
    y='bill_depth_mm',
    color='green',
    hue='sex',
    height=7,
    ratio=5,
    space=0.3,
    kind='kde'
    )
sns.despine()
plt.show()
```
