# KDE plot is just a smoother version of histograms

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

```python
df = sns.load_dataset('penguins')
sdf = df.loc[0:70]
```

```python
sns.kdeplot(
    sdf,
    x='bill_length_mm',
    y='body_mass_g',
    hue='island',
    bw_adjust=0.6,
    fill=True, #NOTE: this is depreciated version shade=True
    multiple='stack'
    )
plt.show()
```
