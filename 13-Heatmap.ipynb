```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

# 🔥 What is a Heatmap? (Super Simple Definition)

A heatmap is a plot that shows data using colors instead of numbers.

Darker / brighter colors = higher values
Lighter / softer colors = lower values

It’s mainly used for correlation, patterns, and matrix data.

# 🧠 Common ML Uses of Heatmap

1)Coorelation matrix : sns.heatmap(df.corr(), annot=True)

2)2️⃣ Confusion Matrix : sns.heatmap(cm, annot=True, fmt='d')       [cm = colormap, fmt = format decimals eg : .2f , .3f]

3)3️⃣ Missing-value visualization 
        sns.heatmap(df.isnull())
Quickly shows which columns have NA values.

# 🛠️ Most Important Parameters (You’ll Use These 95% of the Time)
1)🔶 data=

The matrix you want to show:

sns.heatmap(df)

2)🔶 annot=

Write values inside the boxes.

sns.heatmap(df, annot=True)

3)🔶 fmt=

Format the numbers (useful for correlation):

sns.heatmap(df.corr(), annot=True, fmt='.2f')

4)🔶 cmap=

Color theme (very important for clarity)
Common choices:

'coolwarm'
'viridis'
'magma'
'crest'
'rocket'

sns.heatmap(df.corr(), cmap='coolwarm')

5)🔶 linewidths=

Add small lines between cells (improves clarity)

sns.heatmap(df.corr(), linewidths=0.5)

6)🔶 vmin & vmax

Fix color range across multiple heatmaps.

sns.heatmap(df, vmin=0, vmax=1)

7)🔶 cbar=False

Hide color bar.

# 🔥 Example Heatmap You’ll Use 90% of the Time
plt.figure(figsize=(10,6))
sns.heatmap(df.corr(), annot=True, cmap='coolwarm', fmt='.2f')

# 🧩 Heatmap vs Cluster Heatmap (Bonus Tip)

Seaborn also has:

        sns.clustermap(df)

This automatically clusters similar rows/columns.
Useful for grouping related features.

# 🚫 Limitations of Heatmap
❌ Only for numeric matrices

Categorical data must be encoded first.

❌ Colors can become misleading

If vmin and vmax are not set properly.

❌ Large matrices become messy

Heatmaps get cluttered when columns > 50.

```python
#NOTE : loading data penguins 
dataframe = sns.load_dataset('penguins')
smalldf = dataframe.loc[0:20]
smalldf
```

# NOTE: issue : 

The issue is that your smalldf contains a string column (attnr with values like 'divided' and 'focused'). Heatmaps can only display numeric data, so you need to select only the numeric columns.

```python
column = ["bill_length_mm",	"bill_depth_mm", "flipper_length_mm", "body_mass_g"]

sns.set_context('talk')
sns.heatmap(
    data=smalldf[column].corr(),
    annot=True,
    annot_kws={"color":"black", "fontstyle":"italic", "alpha":0.8},
    fmt='.3f',
    cmap='coolwarm',
   #  NOTE cbar=False, --> This is used to hide the color intensity bar
    linewidths=5,
    linecolor='yellow',
    vmin=0, #NOTE : vmin & vmax is used to set the range of correlation value in the side bar
    vmax=1
    )
plt.show()
```
