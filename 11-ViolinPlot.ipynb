```python
import matplotlib.pyplot as plt 
import seaborn as sns
import pandas as pd
```

```python
#loading datasets of penguins 
ds = sns.load_dataset('penguins')
smallds = ds.iloc[0:48]
smallds
```

```python
sns.violinplot(
    smallds,
    x='island',
    y='bill_length_mm',
    hue='sex',
    split=True,
    inner='quartile'
    )
plt.show()
```

# Using inner = 'stick'

```python
sns.violinplot(
    smallds,
    x='island',
    y='bill_length_mm',
    hue='sex',
    split=True,
    inner='stick'
    )
plt.show()
```

# Using split = 'point'

```python
sns.set_style('darkgrid')
sns.set_context('talk')
sns.violinplot(
    smallds,
    x='island',
    y='bill_length_mm',
    hue='sex',
    split=True,
    inner='point',
    linecolor='black',
    linewidth=4,
    bw_adjust=1,
    cut=0,
    scale='count',
    orient='v'
    )
plt.show()
```

# ✅ 1. What is a Swarmplot? (Simple Definition)

A swarmplot displays every individual data point of a categorical feature and spreads them slightly so they do not overlap.

Think of it as:
🔹 Scatterplot for categories
🔹 Shows actual observations
🔹 No loss of information (unlike boxplot or violin plot)

#   Simply it is : Violin plot + distribution of categorical data points

```python
dataset = sns.load_dataset('penguins')
dataset
smallset = dataset.iloc[0:70]
```

```python
sns.set_style('darkgrid')
sns.set_context('talk')
sns.violinplot(
    smallds,
    x='island',
    y='bill_length_mm',
    hue='sex',
    split=True,
    inner='point',
    linecolor='black',
    linewidth=4,
    bw_adjust=1,
    cut=0,
    density_norm='count',
    orient='v'
    )
sns.swarmplot(
    smallset,
    x='sex',
    y='bill_length_mm',
    hue='island'
    )
plt.show()
```

# 🛠️ Basic KDE plot in Seaborn
sns.kdeplot(data=df, x='age')

🔥 Useful Parameters (you will use these often)
1. shade=True

Fills the area under the curve.

sns.kdeplot(x='age', data=df, shade=True)

2. hue='column'

Separate curves for different groups.

sns.kdeplot(x='age', data=df, hue='gender')

3. multiple='stack' or 'fill'

For overlapping density of multiple groups.

sns.kdeplot(x='income', hue='job', multiple='stack')

4. bw_adjust= (bandwidth)

Controls smoothness.

Small → more wiggly

Large → smoother

sns.kdeplot(x='age', data=df, bw_adjust=0.5)

5. cut=

Controls how far KDE extends beyond data range.

sns.kdeplot(x='salary', data=df, cut=0)
