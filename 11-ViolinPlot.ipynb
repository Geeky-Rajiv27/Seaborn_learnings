```python
import matplotlib.pyplot as plt 
import seaborn as sns
import pandas as pd
```

```python
#loading datasets of penguins 
ds = sns.load_dataset('penguins')
smallds = ds.iloc[0:48]
smallds
```

```python
sns.violinplot(
    smallds,
    x='island',
    y='bill_length_mm',
    hue='sex',
    split=True,
    inner='quartile'
    )
plt.show()
```

# Using inner = 'stick'

```python
sns.violinplot(
    smallds,
    x='island',
    y='bill_length_mm',
    hue='sex',
    split=True,
    inner='stick'
    )
plt.show()
```

# Using split = 'point'

```python
sns.set_style('darkgrid')
sns.set_context('talk')
sns.violinplot(
    smallds,
    x='island',
    y='bill_length_mm',
    hue='sex',
    split=True,
    inner='point',
    linecolor='black',
    linewidth=4,
    bw_adjust=1,
    cut=0,
    scale='count',
    orient='v'
    )
plt.show()
```
