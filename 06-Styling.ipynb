```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
```

```python
dataset = sns.load_dataset('diamonds')
dataset
#now let's work with less rows of data
smallset = dataset.iloc[0:70]
print(smallset)
```

## Learning all about Styling in Seaborn in detail

```python
sns.set_style('whitegrid')
sns.set_context('notebook')
sns.scatterplot(
    smallset,     
    y='depth',
    x='price',
    hue='cut',
    #size='carat',          This cannot be used together with s=
    s=90,        #NOTE size of all dots will be same
    edgecolor='black',
    linewidth=2,
    style='cut'
)
sns.despine(bottom=True)
plt.show()
```
