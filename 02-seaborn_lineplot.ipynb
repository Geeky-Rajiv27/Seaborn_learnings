#  Seaborn in Line plot

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

## Doing using regular matplotlib

```python
var = [1,2,3,4,5,6,7]
var1= [2,3,4,1,5,6,7]
plt.plot(var, var1)
plt.show()
```

# Doing using seaborn :
#               seaborn doesn't supports only (x,y) for line plot as in matplotlib since 
#      seaborn works in datasets so for that we cannot do just an array of numbers to plot
#       the line . We need to import datasets.

```python
df = pd.DataFrame({"var":var, "var_1":var1})
sns.lineplot(data=df, x="var", y="var_1")
#below line contenet i.e;<Axes: xlabel='var', ylabel='var_1'> can be removed using :
plt.show()      #method
```

# Here we can see if we plot a line graph using seaborn the xlabel and ylabel is implitcity
# provided.


# Case : if we don't create a dictionary for dataframes that we gonna need everytime in 
# Seaborn then the output will be same as we get in matplotlib i.e; without xlabel and ylable

# Using datasets provided by Seaborn itself :

```python
datasets = sns.load_dataset('penguins')
print(datasets.head())
```

```python
#Making a graph (lineplot) for bill_length_mm and bill_depth_mm
x = datasets['bill_length_mm']
y = datasets['bill_depth_mm']
sns.lineplot(data=datasets, x=x, y=y)
plt.show()      #--> ;moving unwanted line details of Axes 
```

# hue= name of column from datasets     {legend box creation}

```python
#now adding a new attribute in the lineplot of seaborn (hue=)
# ---> This basically do work of legend box with a very small codes
sns.lineplot(data=datasets, x=x, y=y, hue="sex")
```

# Using attribute: size = integer value

```python
sns.lineplot(data=datasets, x=x, y=y, hue="sex", size=10)
plt.show()
```

# Using  palette : changing colors of the line graph along with their representation inside the hue box

```python
sns.lineplot(data=datasets, x=x, y=y, hue="sex", palette='Accent')
```

# again changing color 

```python
sns.lineplot(data=datasets, x=x, y=y, hue="sex", palette='coolwarm')
```

# Using marker for more clear and distinct representations
# ===> marker is not recommended in lineplot of seaborn since it doesn't support list of makrers in lineplot but marker can be used in scatterplot because it is often prefered

# ALso using (style=)

```python
sns.lineplot(data=datasets, x=x, y=y, hue="sex", palette='Accent', markers=['o', '^', 's'], size=50, style="species")
plt.show()
```

# USing (dashes = True or False)     and (legend = True or False)

```python

sns.lineplot(
    data=datasets,
    x=x,
    y=y,
    hue="sex",
    palette='Accent',
    markers=['o', '^', 's'],
    size=50,
    style="species",
    dashes=[(3,3), (1,3), (None, None)],       #--> dashed for male and (1,3) i.e; dashed dot 
    #       legend=False        --> This line helps to remove the legend box from the graph 
    )

plt.show()
```

# Adding grids and title

```python
sns.lineplot(
    data=datasets,
    x=x,
    y=y,
    hue="sex",
    palette='Accent',
    markers=['o', '^', 's'],
    size=50,
    style="species",
    dashes=[(3,3), (1,3), (None, None)],       #--> dashed for male and (1,3) i.e; dashed dot 
    #       legend=False        --> This line helps to remove the legend box from the graph 
    )
plt.grid()
plt.title('LinePlot using Seaborn')
plt.show()
```
