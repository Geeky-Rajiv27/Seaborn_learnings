```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
```

```python
dataset = sns.load_dataset('penguins')
dataset
#now let's work with less rows of data
smallset = dataset.iloc[0:100]
print(smallset)
```

# The shaded band represents the uncertainty of your current sample's regression line as an estimate of the true population regression line.
The true regression line (the "population line") is a fixed, but unknown, line that describes the relationship between bill_length_mm and bill_depth_mm for all possible penguins of this type (the entire population).
Your sample's regression line (the blue line in the graph) is an estimate of that true line, based only on the data points you collected (your sample).
The 95% confidence level means:
If you were to repeat the entire process—taking new, random samples of the same size many times and calculating a new regression line and a new 95% confidence band for each sample—approximately 95% of those newly generated shaded bands would contain the true population regression line within their boundaries.

```python
sns.regplot(
    dataset,
    x='bill_length_mm',
    y='bill_depth_mm',
    ci=95
    )
plt.show()
```

# In the context of the graph you sent, the "true population" refers to all adult penguins of the species represented in that dataset

# scatter =


```python
sns.regplot(
    dataset,
    x='bill_length_mm',
    y='bill_depth_mm',
    ci=95,
    scatter=False   # This disables or turn off data points
    )
plt.show()
```

# line_kws = {dictionary}     ---> used for line styling

```python
sns.regplot(
    dataset,
    x='bill_length_mm',
    y='bill_depth_mm',
    ci=95,
    line_kws={"color":"red", "linewidth":2}
    )
plt.title('Linear relation between Bill_length_mm and Bill_depth_mm')
plt.show()
```

# scatter_kws = used for styling dots

```python
sns.set_style('whitegrid')
sns.set_context('talk')
sns.regplot(
    dataset,
    x='bill_length_mm',
    y='bill_depth_mm',
    ci=95,
    line_kws={"color":"red", "linewidth":2},
    scatter_kws={"s":50, "edgecolor":"black", "alpha":1}
    )
sns.despine()
plt.title('Linear relation between Bill_length_mm and Bill_depth_mm')
plt.show()
```
