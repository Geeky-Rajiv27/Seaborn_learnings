```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
```

```python
dataset = sns.load_dataset('diamonds')
dataset
#now let's work with less rows of data
smallset = dataset.iloc[0:100]
print(smallset)
```

```python
sns.set_style('whitegrid')
sns.set_context('notebook')
sns.scatterplot(
    smallset,     
    y='depth',
    x='price',
    hue='cut',
    #size='carat',          This cannot be used together with s=
    s=90,        #NOTE size of all dots will be same
    edgecolor='black',
    linewidth=2,
    style='cut',
    palette='terrain_r'
)
sns.despine(bottom=True)    #NOTE : to remove left spine do (left = True)
plt.show()
```

# 💡 Why we use jitter:

When you plot a categorical feature (like "male" vs "female"), all points sit on the same line, so they overlap.
jitter adds a small random movement left/right so points spread out.

# 💡 What it does:

dodge=True spaces out the hue categories side-by-side instead of stacking them on top of each other.

✅ 3. dodge — move hue groups side-by-side

When you use hue, points overlap.
So dodge=True separates them.

✔ Example:

Without dodge → male & female points overlap under each category.
With dodge → male points shift slightly right & female left, so both are visible.

```python
sns.set_style('whitegrid')
sns.set_context('notebook')
sns.stripplot(
    smallset,     
    y='depth',
    x='price',
    hue='cut',
    #size='carat',          This cannot be used together with s=
    s=10,        #NOTE size of all dots will be same
    edgecolor='black',
    linewidth=2,
    palette='terrain_r',
    dodge=True,
    jitter=True
)
sns.despine(bottom=True)    #NOTE : to remove left spine do (left = True)
plt.show()
```

## -------------------------( SWARMPLOT )-----------------------------------##

🎯 1. What is swarmplot()?

A swarmplot is similar to a stripplot, but points never overlap.

Think of it as:

“A smart and organized stripplot that arranges points neatly so no two points sit on top of each other.”

# ❗ Note:

swarmplot does not support dodge=True in older seaborn versions.
Newer versions allow dodge, but with limitations.

# ❌ 6. No jitter needed

Unlike stripplot, swarmplot automatically avoids overlaps.

# 🎯 3. Basic Syntax
sns.swarmplot(data=df, x="category", y="value")

```python
pawnin = sns.load_dataset('penguins')
pawnin
sns.set_context('talk')
sns.set_style('darkgrid')
sns.swarmplot(
    pawnin,
    x='species',
    y='bill_length_mm',
    hue='island',
    edgecolor='black',
    linewidth=2
    )
sns.despine()
plt.show()
```
