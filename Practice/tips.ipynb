```python
import pandas as pd
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
```

```python
datasets_Names = sns.get_dataset_names()
print(datasets_Names)
```

```python
df = sns.load_dataset('tips')
df
```

```python
#NOTE : Grouping and Aggregration
df_bill = df.groupby(['day', 'time'])['total_bill'].mean().reset_index()
print(df_bill)
```

```python
#NOTE: now visualizing the customers insights
#-->the average total bill by day of the week and time of day.
sns.set_context('paper')
sns.barplot(
    df_bill,
    x='day',
    y='total_bill',
    hue='time'
    )
plt.show()
```

```python
#NOTE-->the average total bill by day of the week and time of day.

df_tip = df.groupby(['day', 'time'])['tip'].mean().reset_index()
df_tip['tip'] = df_tip['tip'].round(2)      #rounding off the decimal part

#visulaizing the data:
sns.set_context('paper')
sns.barplot(
    df_tip,
    x='day',
    y='tip',
    hue='time'
    )
plt.show()
```

# NOTE : Relation ship between tip and customer's gender along with smoker status

```python
smoker_df = df.groupby(['sex', 'smoker'])['tip'].max().reset_index()
#visulaizing the data:
sns.set_context('paper')
sns.barplot(
    smoker_df,
    x='sex',
    y='tip',
    hue='smoker',
    palette='dark'
    )
plt.show()
```

```python
#NOTE : syntax  for controlling the scale of y-axix
'''
plt.ylim(lower_limit, upper_limit)
plt.yticks(list_of_ticks)
plt.yscale('log')
'''

#NOTE : since we have lots of rows so it is almost to find exact no. of outliers 
# in the columns so we can find via statistical formula : from 5 no summary concept

# formuls : maximum = Q1 - 1.5 X IQR      and         minimum = Q1 + 1.5 X IQR
Q1 = df['total_bill'].quantile(0.25)
Q3 = df['total_bill'].quantile(0.75)
IQR = Q3 - Q1
print(f"The IQR(inter quartile range) is : {(IQR).round(2)}")
max = (Q3 + 1.5 * IQR).round(2)
print(f"The max value is : {max}")
min = (Q1 - 1.5 * IQR).round(2)
print(f"The min value is : {min}")
print("\n")

#NOTE : FIve no. summary for 'total_bill'
print(f"The five no. summary of 'total_bill' is : {(df['total_bill'].describe()).round(2)}")
print("\n")

print(f"The five no. summary of 'tips' is : {(df['tip'].describe()).round(2)}")
print("\n")



#NOTE : counting exact no. of outliers in column 'total_bill'
outlier_count = df[(df['total_bill'] > max) | (df['total_bill'] < min)].shape[0]
print(f"The total no. of outliers in column ['total_bill] is : {outlier_count}")
sns.set_style('whitegrid')
sns.set_context('paper')
sns.boxplot(
    data=df,
    y='total_bill',
    palette='bright'
    )
plt.yticks(range(2, 52, 2))

plt.show()
```

# Outliers in case of 'tip'

```python
#NOTE : since we have lots of rows so it is almost to find exact no. of outliers 
# in the columns so we can find via statistical formula : from 5 no summary concept

# formuls : maximum = Q1 - 1.5 X IQR      and         minimum = Q1 + 1.5 X IQR
Q1 = df['tip'].quantile(0.25)
Q3 = df['tip'].quantile(0.75)
IQR = Q3 - Q1
print(f"The IQR(inter quartile range) is : {(IQR).round(2)}")
max = (Q3 + 1.5 * IQR).round(2)
print(f"The max value is : {max}")
min = (Q1 - 1.5 * IQR).round(2)
print(f"The min value is : {min}")
print("\n")

#NOTE : FIve no. summary for 'tip'
print(f"The five no. summary of 'tip' is : {(df['tip'].describe()).round(2)}")
print("\n")

print(f"The five no. summary of 'tips' is : {(df['tip'].describe()).round(2)}")
print("\n")



#NOTE : counting exact no. of outliers in column 'tip'
outlier_count = df[(df['tip'] > max) | (df['tip'] < min)].shape[0]
print(f"The total no. of outliers in column ['tip] is : {outlier_count}")
sns.set_style('whitegrid')
sns.set_context('paper')
sns.boxplot(
    data=df,
    y='tip',
    palette='bright'
    )
plt.yticks(range(0, 10))

plt.show()
```
