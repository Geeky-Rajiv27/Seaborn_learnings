 ### Histograms in Seaborn

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
dataset = sns.load_dataset('penguins')
dataset
```

```python
sns.histplot(dataset, x='bill_length_mm')
plt.show()
```

## HEre we can see the bins(intervals points) are not aligned properly according to the histplots . So we can improve this by using an attribute called (bins=)

```python
#NOTE :bins=        we can also pass list of the bins that has been already given according to out x
sns.histplot(dataset, x='bill_length_mm', binrange=(35, 60))
plt.show()
```

# --> we can also add a line (curve) through those bars using an attribute called KDE(kernel Density Estimate)

```python
#NOTE: kde= True or False
sns.histplot(dataset, x='bill_length_mm', binrange=(35, 69), kde=True)
plt.show()
```

```python
#NOTE : 
sns.histplot(dataset, x='bill_length_mm', binrange=(35, 69), kde=True, color='g')
plt.show()
```

# we can also add a log scale instead of just normal scale . Using log_scale=True or False

```python
sns.histplot(dataset, x='bill_length_mm', kde=True, log_scale=True) # with bins= it make some issue
plt.show()
```

# we can also use an attribute multiplt='stringValue' 
# to control overlapped distributions

```python
sns.histplot(dataset, x='bill_length_mm', binrange=(35, 69), kde=True, color='g', multiple='layer')
plt.show()
```

## Using (stat='')  to make modification in the scale like to show density of data , probability , normal raw counts and percentage

```python
sns.histplot(
    dataset,
    x='bill_length_mm',
    binrange=(35, 69),
    kde=True,
    color='g',
    hue='sex',
    stat="percent" #instead of probability we can use argumnet 'density' to see density of
    # data that are falling under the kde curve similarly ('probability') , ('count')
    )
plt.show()
```

# we know if we are creating a histplot or displot for no.of sibling people have then seaborn 
# might try to give a smooth curve by assuming no of sibling as a range which is not possible so, discrete= True will make that range again to integer like makes range (1 - 1.5) as (1 or 2)

```python
sns.histplot(
    dataset,
    x='bill_length_mm',
    binrange=(35, 69),
    kde=True,
    color='g',
    hue='sex',
    stat="probability" #instead of probability we can use argumnet 'density' to see density of
    # data that are falling under the kde curve
    
    )
plt.show()
```

# --> we can style the bars simply by using an attribute called(element='')
#   --> This basically for styling the bars we have coded in the graph
# a) bars --> draws the rectangular box and adds a color inside it.
# b) step --> draws only perimeter 
# c) poly --> draws the bars in a polygon shape

```python
sns.histplot(
    dataset,
    x='bill_length_mm',
    binrange=(35, 69),
    kde=True,
    color='g',
    hue='sex',
    stat="probability", #instead of probability we can use argumnet 'density' to see density of
    # data that are falling under the kde curve
    element='bars',
    weights=dataset['bill_length_mm']
    )
plt.show()
```
