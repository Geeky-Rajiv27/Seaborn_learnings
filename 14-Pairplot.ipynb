```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

# When you run a pair plot on a DataFrame with 4 numeric columns:

You get:

Scatterplot between every pair of columns
Histogram/KDE on the diagonal
Optional grouping using hue

This helps you spot patterns quickly.

🛠️ Basic Pair Plot Example
sns.pairplot(df)

# 🔥 Most Important Parameters (Used 95% of the Time)
1️⃣ data=
Pass your DataFrame.

2️⃣ hue=
Color by a categorical column (very powerful!)
                                                sns.pairplot(df, hue='species')
This helps identify class separation.

3️⃣ kind=
Type of diagonal and lower/upper plots.
Common values:

a)'scatter' (default)
b)'kde'
c)'hist'
d)'reg'

Example:
sns.pairplot(df, kind='kde')

4️⃣ diag_kind=
Controls what appears in the diagonal.
                        sns.pairplot(df, diag_kind='kde')

5️⃣ palette=
Set custom colors.
                            sns.pairplot(df, hue='species', palette='viridis')

6️⃣ plot_kws=
Customize each small scatterplot.

                                sns.pairplot(df, plot_kws={'alpha':0.5})


```python
dataset = sns.load_dataset('penguins')
smallset = dataset.loc[0:60]
smallset
```

# For scatter plots (kind='scatter', the default for pairplot off-diagonal plots), you can use keys from matplotlib.pyplot.scatter or seaborn.scatterplot, such as:
s: Marker size (e.g., plot_kws={'s': 50}).

marker: Marker style (e.g., plot_kws={'marker': '+'}).

c or color: Marker color.

alpha: Marker transparency (e.g., plot_kws={'alpha': 0.7}).

edgecolor or ec: Color of the marker edge (e.g., plot_kws={'edgecolor': 'w'}).

linewidth or lw: Width of the marker edge (e.g., plot_kws={'linewidth': 0.5}).

# For regression plots (kind='reg', as in lmplot), you can use keys for both the scatter plot and the regression line:
scatter_kws: A nested dictionary for the scatter points themselves.

line_kws: A nested dictionary for the regression line properties (e.g., line_kws={'lw': 5} for line width).

ci: Size of the confidence interval.

# For KDE plots (kind='kde'), you can use keys from seaborn.kdeplot such as:
levels: Number of contour levels.

shade or fill: Whether to shade the area under the KDE curve.

color: Color of the KDE lines/areas.

linewidth: Width of the KDE lines. 

```python
sns.pairplot(
    smallset,
    hue='island',
    kind='kde',
    diag_kind='hist',
    palette='bright',
    plot_kws={"fill":True, "levels":10}
    )
plt.show()
```
