```python
import matplotlib.pyplot as plt 
import seaborn as sns
import pandas as pd
```

```python
dataframe = sns.load_dataset('penguins')
dataframe
```

#   the small circles are the outliers of the boxplot of that particular data 

```python
sns.set_style('darkgrid')
sns.set_context('talk')
smallset = dataframe.iloc[0:70]
sns.boxplot(smallset, x='sex', y='bill_depth_mm', hue='island')
plt.show()
```

```python
#NOTE : finding the outlier of above graph that is seen on the male section
biscoe_pawnin = dataframe.loc[(dataframe['bill_depth_mm'] > 20) & (dataframe['bill_depth_mm'] < 22) & (dataframe['sex'] == "Male")]
print(biscoe_pawnin)
```
